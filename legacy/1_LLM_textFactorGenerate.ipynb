```
from openai import OpenAI
 
client = OpenAI(
    api_key=__import__("os").environ["MOONSHOT_API_KEY"], # 在这里将 MOONSHOT_API_KEY 替换为你从 Kimi 开放平台申请的 API Key
    base_url="https://api.moonshot.cn/v1",
)

sentiment_list = ['股价跌惨了', '请查询一下股东人数', '我希望最近去公司考察', '李总您好！请问最近是否和比亚迪有合作', '请问董秘对王石套现离场怎么看，有什么处罚措施']
completion = client.chat.completions.create(
    model = "moonshot-v1-8k",
    messages = [
        {"role": "system", "content": "你是金融情感分析领域的专家，请为下面句子包含情绪进行打分以列表返回给我，打分范围为1-5，1表示负面，2表示比较负面，3表示中性，4表示比较正面，5表示正面"},
        {"role": "user", "content": f"{sentiment_list}"}
    ],
    temperature = 0.3,
)
 
# 通过 API 我们获得了 Kimi 大模型给予我们的回复消息（role=assistant）
print(completion.choices[0].message.content)
```

```
similarity_dict = {'股东有多少人':'投资者您好，截至2024年12月31日，公司股东人数为10000人', '最近是否和比亚迪有合作呢':'投资者您好，截至2024年12月31日，公司股东人数为10000人', '最近是否和比亚迪有合作':'投资者您好，详细请参阅即将发布的年报'}
completion = client.chat.completions.create(
    model = "moonshot-v1-8k",
    messages = [
        {"role": "system", "content": "你是金融语义理解的专家，请识别字典中每个键值对的文本是否具有语义的相关性，打分范围为1-3，1表示答非所问，2表示回答有一些相关性，3表示正面回答问题，返回列表给我"},
        {"role": "user", "content": f"{similarity_dict}"}
    ],
    temperature = 0.3,
)


print(completion.choices[0].message.content)
```
